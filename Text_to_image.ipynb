```python
!pip install -q diffusers accelerate

import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
from IPython.display import display

# 1. Load Pretrained Stable Diffusion v1.5 Model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")

# 2. Dynamic User Input
user_prompt = input("\nEnter Image Prompt: ")

# 3. Generate & Display Image
print("\nGenerating Image... Please Wait...")
image = pipe(user_prompt).images[0]

display(image)
image.save("generated_image.png")
print("\nImage Generated Successfully and saved as generated_image.png!")
```

```python
import gradio as gr

def generate_image_gradio(prompt):
    # Re-use the existing 'pipe' from the previous cell
    # pipe is already on 'cuda' and loaded from 'runwayml/stable-diffusion-v1-5'

    # Generate image
    image = pipe(prompt).images[0]
    return image

# Create the Gradio interface
iface = gr.Interface(
    fn=generate_image_gradio,
    inputs=gr.Textbox(lines=2, placeholder="Enter your image prompt here..."),
    outputs=gr.Image(type="pil"),
    title="Stable Diffusion Image Generator",
    description="Generate images using a Stable Diffusion v1.5 model. Enter a prompt and click Submit."
)

# Launch the interface
iface.launch(debug=True)
```
